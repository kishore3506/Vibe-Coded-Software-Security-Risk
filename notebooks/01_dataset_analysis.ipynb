# 01. Dataset Analysis & Exploratory Data Analysis (EDA)
## Project: Artificial Neural Network-Based Detection and Classification of Security Risks in Vibe-Coded Software

This notebook performs comprehensive exploratory data analysis on the **Microsoft CodeXGLUE Defect Detection (Devign benchmark)** dataset.
It inspects source-code samples, label distributions, missing values, duplicates, and code length distributions between secure and vulnerable functions.

```python
import os
import json
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset metadata
meta_path = '../data/dataset_metadata.json'
with open(meta_path, 'r', encoding='utf-8') as f:
    metadata = json.load(f)
print('Dataset Name:', metadata['dataset_name'])
print('Source URL:', metadata['source_url'])
print('Original Total Samples:', metadata['total_original_samples'])
```

```python
# Load train, validation, and test subsets
train_df = pd.read_parquet('../data/processed/train_subset.parquet')
val_df = pd.read_parquet('../data/processed/val_subset.parquet')
test_df = pd.read_parquet('../data/processed/test_subset.parquet')

print('Train Shape:', train_df.shape)
print('Validation Shape:', val_df.shape)
print('Test Shape:', test_df.shape)
train_df.head(3)
```

```python
# Target label distribution
print('Train Class Distribution:')
print(train_df['target'].value_counts(normalize=True))

fig, ax = plt.subplots(figsize=(6, 4))
train_df['target'].value_counts().plot(kind='bar', ax=ax, color=['#4CAF50', '#F44336'])
ax.set_xticklabels(['Secure (0)', 'Vulnerable (1)'], rotation=0)
ax.set_title('Training Split Target Distribution')
ax.set_ylabel('Sample Count')
plt.tight_layout()
plt.show()
```

```python
# Source code length analysis
train_df['char_len'] = train_df['func'].apply(len)
train_df['line_count'] = train_df['func'].apply(lambda x: len(x.splitlines()))

print('Code Length Summary:')
print(train_df.groupby('target')[['char_len', 'line_count']].describe())
```
