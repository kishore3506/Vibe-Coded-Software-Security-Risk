# 03. Artificial Neural Network (ANN) Model Architecture & Training
## Deep Multi-Layer Perceptron (MLP) with Batch Normalization, Dropout, and EarlyStopping

### Architecture:
- **Input Layer:** 85 Features
- **Dense Layer 1:** 128 neurons, ReLU activation
- **Batch Normalization 1 + Dropout (0.3)**
- **Dense Layer 2:** 64 neurons, ReLU activation
- **Batch Normalization 2 + Dropout (0.2)**
- **Dense Layer 3:** 32 neurons, ReLU activation + Dropout (0.2)
- **Output Layer:** 1 neuron, Sigmoid activation (Binary Cross-Entropy loss)

```python
import sys
sys.path.append('..')
from src.train_ann import build_ann_model
from src.data_preprocessing import load_preprocessed_data

X_train, y_train, X_val, y_val, X_test, y_test, feature_names = load_preprocessed_data()
model = build_ann_model(input_dim=X_train.shape[1])
model.summary()
```

```python
from tensorflow import keras
from sklearn.metrics import classification_report, confusion_matrix

# Load trained model and evaluate on test set
best_model = keras.models.load_model('../models/ann_model.keras')
y_prob = best_model.predict(X_test).flatten()
y_pred = (y_prob >= 0.5).astype(int)

print('Test Classification Report:')
print(classification_report(y_test, y_pred, target_names=['Secure (0)', 'Vulnerable (1)']))
```
